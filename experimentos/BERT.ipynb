```python
from models.bert_model import *
```

```python

config = {
    'vocab_size': 30522, 'max_position_embeddings': 512, 'type_vocab_size': 2,
    'hidden_size': 128, 'num_heads': 2, 'num_layers': 2,
    'intermediate_size': 512, 'dropout_rate': 0.1
}

# Construir el modelo
bert_from_scratch = BertModel(config, name='bert')

# Forzar la creación de los pesos
input_ids = tf.random.uniform(shape=(1, 128), dtype=tf.int32, maxval=config['vocab_size'])
segment_ids = tf.zeros((1, 128), dtype=tf.int32)
_ = bert_from_scratch([input_ids, segment_ids])
```

```python
bert_from_scratch.summary()
```

```python
# Comprobar los nombres de las variables (ahora usando .path)
print("\nAlgunos nombres de variables del modelo Keras (usando .path):")
for i in range(5):
    print(bert_from_scratch.weights[i].path)
```

```python
# Cargar los pesos
print("\nCargando pesos del checkpoint...")
checkpoint_path = 'BERT_tiny/bert_model.ckpt'
load_weights_from_checkpoint(bert_from_scratch, checkpoint_path)
```

```python
# Verificar que se cargó un peso correctamente
print("\nVerificación de carga:")
reader = tf.train.load_checkpoint(checkpoint_path)
ckpt_emb = reader.get_tensor('bert/embeddings/word_embeddings')
model_emb = bert_from_scratch.embeddings.word_embeddings.weights[0].numpy()

print("\n¿Los embeddings de token coinciden?", np.allclose(ckpt_emb, model_emb))
```
